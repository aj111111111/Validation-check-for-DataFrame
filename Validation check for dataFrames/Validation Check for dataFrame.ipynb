```python
import pandas as pd
import numpy as np

def fun_remove_invalid_character(df, *col_names):
    """
    Removes alphabets, /, \, and - from multiple specified columns.
    """
    for col_name in col_names:
        if col_name in df.columns:
            # Remove letters, forward slashes, backslashes, and hyphens
            df[col_name] = df[col_name].astype(str).str.replace(r'[A-Za-z/\\-]', '', regex=True)
            
            # Convert empty strings or spaces left behind back to NaN
            df[col_name] = df[col_name].replace(r'^\s*$', np.nan, regex=True)
            
            print(f"Cleaned column: '{col_name}'")
        else:
            print(f"Warning: Column '{col_name}' not found in DataFrame.")
            
    return df

```

```python
for col in df.columns:
    # 1. Convert to string, strip hidden spaces, and try converting to numeric
    converted = pd.to_numeric(df[col].astype(str).str.strip(), errors='coerce')
    
    # 2. If it actually found numbers (and isn't just a pure text column like names), apply it
    if converted.notna().sum() > 0 and df[col].astype(str).str.strip().eq('').sum() == 0:
        # Check if it has decimals to decide int vs float
        df[col] = converted

# Check results
print(df.info())
```
